### 2-3) 데이터 분할

- **Train:** 모델이 학습하는 데 사용하는 데이터. 모델은 이 데이터로 패턴을 배움
- **Validation:** 학습 과정에서 모델 성능을 점검하고 하이퍼파라미터를 조정하는 데 사용하는 데이터. 모델 선택 및 하이퍼파라미터 최적화의 기준이 되며, 최종 평가에는 쓰지 않음
- **Test:** 학습이 모두 끝난 후, 모델의 최종 성능을 평가하는 데 사용하는 데이터. 학습 및 최적화 과정에서 한 번도 보지 않은 데이터로 일반화 성능을 측정함

```python
from sklearn.model_selection import train_test_split

# 입력변수와 출력변수로 분할
X_reg = Regression.drop(["y"], axis=1)
Y_reg = Regression["y"]

X_cla = Classification.drop(["Y"], axis=1)
Y_cla = Classification["Y"]
```

```python
# [회귀] 학습 데이터와 테스트 데이터로 분할

# test_size는 전체 데이터 크기 중 test 데이터의 크기를 얼마의 비중으로 할 것이냐를 뜻함
# random_state는 난수를 고정해주는 함수로, 동일한 값에서는 동일한 데이터가 뽑힘

X_reg_train, X_reg_test, Y_reg_train, Y_reg_test = train_test_split(
    X_reg, Y_reg, test_size=0.3, random_state=42)

print(Regression.shape, X_reg_train.shape, X_reg_test.shape, Y_reg_train.shape, Y_reg_test.shape)
X_reg_train.head(n=5)
```

```python
# [분류] 학습 데이터와 테스트 데이터로 분할
# stratify=Y 옵션은 클래스 비율에 맞게 학습 데이터와 테스트 데이터를 분할해 주는 옵션

X_cla_train, X_cla_test, Y_cla_train, Y_cla_test = train_test_split(
    X_cla, Y_cla, test_size=0.3, random_state=42, stratify=Y_cla)

print(Classification.shape, X_cla_train.shape, X_cla_test.shape, Y_cla_train.shape, Y_cla_test.shape)
X_cla_train.head(n=5)
```

1. **Min-Max Scaling:** 모든 값을 0~1 범위로 변환 (이상치가 적은 데이터에 적합)
2. **Z-score Scaling:** 평균 0, 표준편차 1이 되도록 변환 (정규분포에 가까운 데이터에 적합)

### 3-3) 성능 평가

```python
from sklearn.metrics import (mean_absolute_error, mean_squared_error,
                             accuracy_score, recall_score, precision_score,
                             f1_score, roc_auc_score, confusion_matrix)

# [회귀 성능지표] MAE, MSE, MAPE를 계산해 딕셔너리로 반환
def get_reg_score(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)                   # 평균 절대 오차
    mse = mean_squared_error(y_true, y_pred)                    # 평균 제곱 오차
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100    # 평균 절대 백분율 오차(%)
    return {"MAE": mae, "MSE": mse, "MAPE(%)": mape}

# [분류 성능지표] 혼동행렬에서 지표를 뽑아 딕셔너리로 반환
def get_cla_score(y_true, y_pred, y_prob):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()   # 혼동행렬 4칸 분해
    recall = recall_score(y_true, y_pred)         # 재현율(민감도): 실제 양성 중 맞춘 비율
    specificity = tn / (tn + fp)                  # 특이도: 실제 음성 중 맞춘 비율
    return {
        "Accuracy": accuracy_score(y_true, y_pred),    # 정확도
        "Recall": recall,                               # 재현율
        "Precision": precision_score(y_true, y_pred),   # 정밀도
        "Specificity": specificity,                     # 특이도
        "F1": f1_score(y_true, y_pred),                 # 정밀도·재현율 조화평균
        "G-Mean": np.sqrt(recall * specificity),        # 재현율·특이도 기하평균(불균형 대응)
        "AUC": roc_auc_score(y_true, y_prob),           # ROC 곡선 아래 면적
    }
```

```python
# 성능 평가 - 회귀
# 모델별 클래스 예측값과 양성 확률로 성능지표를 산출해 표로 정리 (MAE 기준 오름차순 정렬 (오차 작은 순))

reg_results = {}
for name in reg_models:
    reg_results[name] = get_reg_score(Y_reg_test_final, reg_preds[name])

reg_score_df = pd.DataFrame(reg_results).T.round(4).sort_values("MAE")
reg_score_df
```

```python
# 성능 평가 - 분류
# 모델별 클래스 예측값과 양성 확률로 성능지표를 산출해 표로 정리 (F1 기준 내림차순 정렬 (성능 좋은 순))

cla_results = {}
for name in cla_models:
    cla_results[name] = get_cla_score(Y_cla_test_final, cla_preds[name], cla_probs[name])

cla_score_df = pd.DataFrame(cla_results).T.round(4).sort_values("F1", ascending=False)
cla_score_df
```

```python
# [방법] IQR 기준 이상치 개수 요약 → 이상치 많은 변수만 boxplot으로 확인
# 1) 변수별 이상치 개수·비율을 한눈에 비교 (막대그래프)
# 2) 상위 변수만 boxplot으로 분포·이상치 위치 확인

def count_iqr_outliers(series):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()

def summarize_outliers(df, target_col, title):
    num_cols = df.select_dtypes(include="number").columns.drop(target_col)

    outlier_summary = pd.DataFrame({
        "count": [count_iqr_outliers(df[c]) for c in num_cols],
        "ratio(%)": [count_iqr_outliers(df[c]) / len(df) * 100 for c in num_cols],
    }, index=num_cols).sort_values("count", ascending=False)

    print(f"\n=== {title} : IQR 이상치 상위 10개 ===")
    display(outlier_summary.head(10).round(2))

    # 변수별 이상치 개수 막대그래프
    plt.figure(figsize=(12, 8))
    outlier_summary["count"].sort_values().plot(kind="barh")
    plt.title(f"{title} - 변수별 IQR 이상치 개수")
    plt.xlabel("이상치 개수")
    plt.tight_layout()
    plt.show()

    # 이상치가 많은 상위 6개 변수만 boxplot 확인
    top_cols = outlier_summary.head(6).index
    n_cols = 3
    n_rows = int(np.ceil(len(top_cols) / n_cols))

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(14, 4 * n_rows))
    axes = np.array(axes).ravel()

    for ax, col in zip(axes, top_cols):
        sns.boxplot(y=df[col], ax=ax, color="skyblue")
        ax.set_title(col)

    for ax in axes[len(top_cols):]:
        ax.axis("off")

    plt.suptitle(f"{title} - 이상치 상위 변수 Boxplot", y=1.02)
    plt.tight_layout()
    plt.show()

    return outlier_summary

reg_outlier_summary = summarize_outliers(Regression, "y", "회귀 데이터")
cla_outlier_summary = summarize_outliers(Classification, "Y", "분류 데이터")
```

**[과제 2]** 지수는 분류 데이터에서 target(Y)과 관련 있어 보이는 변수들을 추려내고 싶습니다. 그런데 변수가 많다 보니 하나씩 비교해보는 게 맞는 방법인지 확신이 서지 않습니다. 효율적으로 비교할 방법을 AI와 의논해서 정하고, 관련성이 높아 보이는 변수가 무엇인지 정리해 보세요.

```python
# [방법] Heatmap(변수 전체 관계) 대신, Y와의 관련성만 추려서 비교
# ① Target 상관계수 정렬 막대그래프 (선형 관계)
# ② Mutual Information (선형+비선형 관계까지 포착)
# ③ 상위 변수만 Boxplot으로 분포 차이 직접 확인

from sklearn.feature_selection import mutual_info_classif

feature_cols = Classification.columns.drop(["Y"])
X = Classification[feature_cols]
y = Classification["Y"]

# ① Target과의 상관계수 (Y가 이진이므로 point-biserial correlation과 동일)
target_corr = X.corrwith(y).sort_values(key=np.abs, ascending=False)

# ② Mutual Information (비선형 관계도 포착, random_state로 재현성 고정)
mi = pd.Series(
    mutual_info_classif(X, y, random_state=42),
    index=feature_cols,
).sort_values(ascending=False)

relevance_summary = pd.DataFrame({
    "corr_with_Y": target_corr,
    "abs_corr": target_corr.abs(),
    "mutual_info": mi,
}).sort_values("mutual_info", ascending=False)

print("=== Y와 관련성 높은 변수 Top 10 (Mutual Info 기준) ===")
display(relevance_summary.head(10).round(4))

# 막대그래프로 한눈에 비교 (상관계수 vs Mutual Info)
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

relevance_summary["abs_corr"].sort_values().plot(kind="barh", ax=axes[0], color="steelblue")
axes[0].set_title("변수별 |Y와의 상관계수|")
axes[0].set_xlabel("|correlation|")

relevance_summary["mutual_info"].sort_values().plot(kind="barh", ax=axes[1], color="darkorange")
axes[1].set_title("변수별 Mutual Information with Y")
axes[1].set_xlabel("mutual information")

plt.tight_layout()
plt.show()

# ③ 관련성 상위 6개 변수는 Y=0/1 분포가 실제로 다른지 Boxplot으로 확인
top_cols = relevance_summary.head(6).index
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.ravel()

for ax, col in zip(axes, top_cols):
    sns.boxplot(data=Classification, x="Y", y=col, hue="Y", legend=False, ax=ax)
    ax.set_title(col)

plt.suptitle("Y와 관련성 높은 변수 - Boxplot (Y=0 vs Y=1)", y=1.02)
plt.tight_layout()
plt.show()

relevance_summary
```

**[과제 3]** 지수는 모델에 같이 넣었을 때 문제가 될 만큼 상관관계가 높은 변수쌍이 있는지 궁금합니다. 히트맵을 눈으로 훑어보는 것만으로는 놓치는 게 있을 것 같아, 더 체계적으로 찾아낼 방법이 없을지 AI와 상의해 보세요.

```python
# [방법] 절대 기준(고정 threshold)으로 다중공선성 후보를 체계적으로 추출
# ① 상관계수 |r| >= 0.8 인 변수쌍만 필터링 (정렬)
# ② VIF >= 10 인 변수만 필터링 (정렬)
# → 둘 다 해당하는 변수가 가장 확실한 다중공선성 후보

from statsmodels.stats.outliers_influence import variance_inflation_factor

CORR_THRESHOLD = 0.8
VIF_THRESHOLD = 10

def get_high_corr_pairs(df, exclude_cols, threshold=CORR_THRESHOLD):
    num_df = df.drop(columns=exclude_cols)
    corr_matrix = num_df.corr()

    pairs = []
    cols = corr_matrix.columns
    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):
            r = corr_matrix.iloc[i, j]
            if abs(r) >= threshold:
                pairs.append((cols[i], cols[j], r))

    pairs_df = pd.DataFrame(pairs, columns=["변수1", "변수2", "corr"])
    return pairs_df.reindex(pairs_df["corr"].abs().sort_values(ascending=False).index)

def get_high_vif(df, exclude_cols, threshold=VIF_THRESHOLD):
    num_df = df.drop(columns=exclude_cols).dropna()
    vif_df = pd.DataFrame({
        "변수": num_df.columns,
        "VIF": [variance_inflation_factor(num_df.values, i) for i in range(num_df.shape[1])],
    }).sort_values("VIF", ascending=False)
    return vif_df[vif_df["VIF"] >= threshold]

def check_multicollinearity(df, target_col, title):
    print(f"\n{'='*10} {title} {'='*10}")

    high_corr = get_high_corr_pairs(df, exclude_cols=[target_col])
    print(f"\n[상관계수 |r| >= {CORR_THRESHOLD}] 인 변수쌍: {len(high_corr)}개")
    display(high_corr)

    high_vif = get_high_vif(df, exclude_cols=[target_col])
    print(f"\n[VIF >= {VIF_THRESHOLD}] 인 변수: {len(high_vif)}개")
    display(high_vif)

    return high_corr, high_vif

reg_high_corr, reg_high_vif = check_multicollinearity(Regression, "y", "회귀 데이터")
cla_high_corr, cla_high_vif = check_multicollinearity(Classification.drop(columns=["ID"], errors="ignore"), "Y", "분류 데이터")
```

### 5-2) 전처리 심화

**[과제 4]** 지수는 수치형 변수들의 스케일이 제각각이라 모델 학습 전에 스케일링이 필요하다고 판단했습니다. 그런데 Min-Max와 Standard 중 어떤 걸 써야 할지, 변수마다 다르게 적용해야 하는 건 아닌지 확신이 서지 않습니다. AI와 어떤 스케일링이 적합할지 논의하고, 정한 방식을 적용해 보세요.

**[과제 5]** 지수는 앞서(5-1) 상관관계가 높은 변수쌍 후보를 찾아뒀습니다. 그런데 막상 "상관계수가 몇 이상이면 지워야 하는지", "쌍 중에서 어떤 변수를 남겨야 하는지"는 정해진 답이 없어 고민이 됩니다. threshold를 어떻게 잡을지, 어떤 변수를 남길지 AI와 함께 근거를 세워서 정하고, 정한 기준대로 제거한 뒤 모델 성능이 어떻게 달라지는지 확인해 보세요.

```python
# [방법] Stepwise VIF 제거
# - threshold: VIF >= 10 (표준 컷오프)
# - 매 반복마다 VIF가 가장 높은 변수를 제거하고 다시 계산 (한 번에 여러 개 제거 X)
# - VIF가 서로 근접한(오차 5% 이내) 변수들 중에서는 target 관련성(|corr|)이 낮은 쪽을 제거
# - (변수는 z-score로 표준화된 상태로 가정)
# - 제거 전/후 모델 성능을 비교해 실제 효과 확인

from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, accuracy_score, roc_auc_score

VIF_THRESHOLD = 10

def stepwise_vif_elimination(X_scaled, relevance, threshold=VIF_THRESHOLD, tol=0.05, verbose=True):
    features = list(X_scaled.columns)
    removed_log = []

    while True:
        vif = pd.Series(
            [variance_inflation_factor(X_scaled[features].values, i) for i in range(len(features))],
            index=features,
        )
        max_vif = vif.max()
        if max_vif < threshold:
            break

        # 최고 VIF와 근접한(5% 이내) 변수들 중 target 관련성이 가장 낮은 변수를 제거
        near_max = vif[vif >= max_vif * (1 - tol)].index
        drop_col = relevance[near_max].idxmin()

        removed_log.append({"제거변수": drop_col, "VIF": round(vif[drop_col], 2), "relevance": round(relevance[drop_col], 4)})
        if verbose:
            print(f"제거: {drop_col:>6}  (VIF={vif[drop_col]:.2f}, relevance={relevance[drop_col]:.4f})")

        features.remove(drop_col)

    return features, pd.DataFrame(removed_log)


# ------------------------------------------------------------------
# [회귀] Regression 데이터
# ------------------------------------------------------------------
X_reg = Regression.drop(columns=["y"])
y_reg = Regression["y"]

X_reg_scaled = pd.DataFrame(StandardScaler().fit_transform(X_reg), columns=X_reg.columns)
relevance_reg = X_reg.corrwith(y_reg).abs()

print("=== [회귀] Stepwise VIF 제거 ===")
reg_final_features, reg_removed_log = stepwise_vif_elimination(X_reg_scaled, relevance_reg)
print(f"\n제거된 변수 수: {len(reg_removed_log)} / 남은 변수 수: {len(reg_final_features)}")
display(reg_removed_log)

# 제거 전/후 성능 비교 (LinearRegression, MAE 기준)
Xtr_full, Xte_full, ytr, yte = train_test_split(X_reg_scaled, y_reg, test_size=0.3, random_state=42)
Xtr_sel, Xte_sel = Xtr_full[reg_final_features], Xte_full[reg_final_features]

mae_before = mean_absolute_error(yte, LinearRegression().fit(Xtr_full, ytr).predict(Xte_full))
mae_after = mean_absolute_error(yte, LinearRegression().fit(Xtr_sel, ytr).predict(Xte_sel))

print("\n[회귀 성능 비교 - MAE]")
print(f"제거 전 (변수 {X_reg_scaled.shape[1]}개): {mae_before:.4f}")
print(f"제거 후 (변수 {len(reg_final_features)}개): {mae_after:.4f}")


# ------------------------------------------------------------------
# [분류] Classification 데이터
# ------------------------------------------------------------------
X_cla = Classification.drop(columns=["Y"])
y_cla = Classification["Y"]

X_cla_scaled = pd.DataFrame(StandardScaler().fit_transform(X_cla), columns=X_cla.columns)
relevance_cla = X_cla.corrwith(y_cla).abs()

print("\n=== [분류] Stepwise VIF 제거 ===")
cla_final_features, cla_removed_log = stepwise_vif_elimination(X_cla_scaled, relevance_cla)
print(f"\n제거된 변수 수: {len(cla_removed_log)} / 남은 변수 수: {len(cla_final_features)}")
display(cla_removed_log)

# 제거 전/후 성능 비교 (LogisticRegression, Accuracy·AUC 기준)
Xtr_full, Xte_full, ytr, yte = train_test_split(X_cla_scaled, y_cla, test_size=0.3, random_state=42, stratify=y_cla)
Xtr_sel, Xte_sel = Xtr_full[cla_final_features], Xte_full[cla_final_features]

model_before = LogisticRegression(max_iter=1000).fit(Xtr_full, ytr)
model_after = LogisticRegression(max_iter=1000).fit(Xtr_sel, ytr)

acc_before = accuracy_score(yte, model_before.predict(Xte_full))
auc_before = roc_auc_score(yte, model_before.predict_proba(Xte_full)[:, 1])
acc_after = accuracy_score(yte, model_after.predict(Xte_sel))
auc_after = roc_auc_score(yte, model_after.predict_proba(Xte_sel)[:, 1])

print("\n[분류 성능 비교 - Accuracy / AUC]")
print(f"제거 전 (변수 {X_cla_scaled.shape[1]}개): Accuracy={acc_before:.4f}, AUC={auc_before:.4f}")
print(f"제거 후 (변수 {len(cla_final_features)}개): Accuracy={acc_after:.4f}, AUC={auc_after:.4f}")
```

### 5-4) 모델링 고도화 심화

**[과제 7]** 지수는 모델 성능을 더 끌어올리고 싶은데, 직접 k-fold cross validation으로 하나하나 비교해봐야 할지, GridSearchCV로 자동 탐색해야 할지, 아니면 RandomizedSearchCV가 더 나을지 헷갈립니다. 각 방법이 어떤 상황에 적합한지 AI와 상의해서 이번 데이터와 모델에 맞는 방법을 정하고, 정한 방법으로 직접 탐색해 보세요.

```python
# [방법] 모델 특성에 따라 하이퍼파라미터 탐색 방법을 다르게 선택
# - [회귀] Ridge: 하이퍼파라미터가 alpha 1개뿐 → GridSearchCV로 전체 탐색
# - [분류] CatBoost: 하이퍼파라미터가 많고 상호작용 있음 → RandomizedSearchCV로 효율 탐색

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, accuracy_score, roc_auc_score
from catboost import CatBoostClassifier

# ------------------------------------------------------------------
# [회귀] Ridge - GridSearchCV
# ------------------------------------------------------------------
X_reg = Regression.drop(columns=["y"])
y_reg = Regression["y"]

Xtr, Xte, ytr, yte = train_test_split(X_reg, y_reg, test_size=0.3, random_state=42)
scaler_reg = StandardScaler().fit(Xtr)
Xtr_s, Xte_s = scaler_reg.transform(Xtr), scaler_reg.transform(Xte)

mae_default = mean_absolute_error(yte, Ridge().fit(Xtr_s, ytr).predict(Xte_s))

ridge_grid = GridSearchCV(
    Ridge(),
    param_grid={"alpha": np.logspace(-3, 3, 13)},  # 0.001 ~ 1000, 13개 값 전체 탐색
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1,
)
ridge_grid.fit(Xtr_s, ytr)
mae_tuned = mean_absolute_error(yte, ridge_grid.best_estimator_.predict(Xte_s))

print("=== [회귀] Ridge GridSearchCV 결과 ===")
print(f"최적 alpha: {ridge_grid.best_params_['alpha']:.4f}  (CV MAE: {-ridge_grid.best_score_:.4f})")
print(f"기본(alpha=1.0) 테스트 MAE : {mae_default:.4f}")
print(f"튜닝 후 테스트 MAE        : {mae_tuned:.4f}")


# ------------------------------------------------------------------
# [분류] CatBoost - RandomizedSearchCV
# ------------------------------------------------------------------
X_cla = Classification.drop(columns=["Y"])
y_cla = Classification["Y"]

Xtr, Xte, ytr, yte = train_test_split(X_cla, y_cla, test_size=0.3, random_state=42, stratify=y_cla)
scaler_cla = StandardScaler().fit(Xtr)
Xtr_s, Xte_s = scaler_cla.transform(Xtr), scaler_cla.transform(Xte)

cat_default = CatBoostClassifier(verbose=0, random_state=42).fit(Xtr_s, ytr)
acc_default = accuracy_score(yte, cat_default.predict(Xte_s))
auc_default = roc_auc_score(yte, cat_default.predict_proba(Xte_s)[:, 1])

cat_random = RandomizedSearchCV(
    CatBoostClassifier(verbose=0, random_state=42),
    param_distributions={
        "depth": [4, 6, 8, 10],
        "learning_rate": np.linspace(0.01, 0.3, 10),
        "iterations": [100, 200, 300, 500],
        "l2_leaf_reg": [1, 3, 5, 7, 9],
    },
    n_iter=20,        # 넓은 공간에서 20번만 무작위 샘플링
    cv=3,
    scoring="roc_auc",
    random_state=42,
    n_jobs=-1,
)
cat_random.fit(Xtr_s, ytr)
acc_tuned = accuracy_score(yte, cat_random.best_estimator_.predict(Xte_s))
auc_tuned = roc_auc_score(yte, cat_random.best_estimator_.predict_proba(Xte_s)[:, 1])

print("\n=== [분류] CatBoost RandomizedSearchCV 결과 ===")
print(f"최적 하이퍼파라미터: {cat_random.best_params_}  (CV AUC: {cat_random.best_score_:.4f})")
print(f"기본 파라미터 테스트   : Accuracy={acc_default:.4f}, AUC={auc_default:.4f}")
print(f"튜닝 후 테스트         : Accuracy={acc_tuned:.4f}, AUC={auc_tuned:.4f}")
```
